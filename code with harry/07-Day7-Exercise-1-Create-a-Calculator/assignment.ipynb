```python
print("This is a calculator")

def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def div(a, b):
    return a / b

def calculator(a, b, x):
    match x:
        case 1:
            print(f"Addition of {a} & {b} is {add(a, b)}")
        case 2:
            print(f"Subtraction of {a} & {b} is {sub(a, b)}")
        case 3:
            print(f"Multiplication of {a} & {b} is {mul(a, b)}")
        case 4:
            print(f"Division of {a} & {b} is {div(a, b)}")

while True:
    # Step 1: Input two numbers
    a = int(input("Enter the first number: "))
    b = int(input("Enter the second number: "))

    while True:
        # Step 2: Choose an operation
        print("1. Add  2. Subtract  3. Multiply  4. Divide")
        x = int(input("Enter the operation you want to perform: "))

        # Validate the operation
        while x > 4 or x < 1:
            print(f"{x} is not a valid opereation , please enter a valid operation")
            x = int(input("Enter the operation you want to perform: "))

        # Perform the operation
        calculator(a, b, x)

        # Step 3: Ask if the user wants to continue
        print("Do you want to:")
        print("1. Perform another operation on the same values")
        print("2. Perform operation on another set of values")
        print("3. Exit the calculator")
        choice = int(input("Enter your choice (1/2/3): "))

        if choice == 1:
            continue  # Re-loop to perform another operation on the same values
        elif choice == 2:
            break  # Break the inner loop to input new values
        elif choice == 3:
            print("Exiting the calculator. Goodbye!")
            exit()  # Exit the program
        else:
            print("Invalid choice. Please try again.")
```
